# 10.8 Python Errors

## Syntax Error

Compile (Translation) error, that is caught before the program runs. In other words, it stops the program from being able to run.

## Exception (Run-time Error)

The syntax is fine (it is legal), so the programs runs, however, because of this Exception, the program will stop and give you an error.

## Logic Error

Type of run-time error. The program runs normally gives you output as it is suppose to. The problem is, if you aren't paything attention, the output is INCORRECT. Ex. print(2 + 2)   OUTPUT: 28 

Logic Errors can be difficult to spot. This is why you have to have an understanding of what the output should be.
